```python
#!/usr/bin/env python3
"""
SimpleHydrate - very simple water intake tracker using only Python basics.

Files used:
 - entries.csv  (created automatically)
 - profile.json (created automatically)

Usage: python water_tracker.py
"""

import csv
import json
import os
from datetime import datetime, date

ENTRIES_FILE = "entries.csv"
PROFILE_FILE = "profile.json"

# ----------------- Utilities -----------------

def ensure_files():
    if not os.path.exists(ENTRIES_FILE):
        with open(ENTRIES_FILE, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(["id", "date", "time", "amount_ml", "note"])
    if not os.path.exists(PROFILE_FILE):
        profile = {"username": "user", "daily_goal_ml": 2000}
        with open(PROFILE_FILE, "w") as f:
            json.dump(profile, f)

def load_profile():
    with open(PROFILE_FILE, "r") as f:
        return json.load(f)

def save_profile(profile):
    with open(PROFILE_FILE, "w") as f:
        json.dump(profile, f)

def read_entries():
    entries = []
    with open(ENTRIES_FILE, newline="") as f:
        reader = csv.DictReader(f)
        for r in reader:
            r["id"] = int(r["id"])
            r["amount_ml"] = int(r["amount_ml"])
            entries.append(r)
    return entries

def write_entries(entries):
    with open(ENTRIES_FILE, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["id", "date", "time", "amount_ml", "note"])
        for e in entries:
            writer.writerow([e["id"], e["date"], e["time"], e["amount_ml"], e.get("note","")])

def next_id(entries):
    if not entries:
        return 1
    return max(e["id"] for e in entries) + 1

# ----------------- Features -----------------

def add_entry():
    try:
        amt = input("Enter amount in ml (e.g., 200): ").strip()
        if not amt.isdigit():
            print("Please enter a numeric value.")
            return
        amt = int(amt)
        note = input("Optional note: ").strip()
        now = datetime.now()
        entries = read_entries()
        eid = next_id(entries)
        entries.append({
            "id": eid,
            "date": now.strftime("%Y-%m-%d"),
            "time": now.strftime("%H:%M:%S"),
            "amount_ml": amt,
            "note": note
        })
        write_entries(entries)
        print(f"Added {amt} ml (id={eid}) at {now.strftime('%H:%M:%S')}")
    except Exception as ex:
        print("Error adding entry:", ex)

def list_entries(for_date=None):
    entries = read_entries()
    if for_date is None:
        for_date = date.today().strftime("%Y-%m-%d")
    print(f"Entries for {for_date}:")
    total = 0
    found = False
    for e in entries:
        if e["date"] == for_date:
            found = True
            print(f"  id={e['id']} time={e['time']} {e['amount_ml']} ml note={e.get('note','')}")
            total += e["amount_ml"]
    if not found:
        print("  (no entries)")
    print("Total:", total, "ml")
    profile = load_profile()
    print("Goal:", profile.get("daily_goal_ml", 0), "ml — Progress:", f"{total}/{profile.get('daily_goal_ml',0)}")

def edit_entry():
    entries = read_entries()
    eid = input("Enter entry id to edit: ").strip()
    if not eid.isdigit():
        print("Invalid id.")
        return
    eid = int(eid)
    for e in entries:
        if e["id"] == eid:
            print("Current:", e)
            new_amt = input(f"New amount in ml (enter to keep {e['amount_ml']}): ").strip()
            if new_amt:
                if not new_amt.isdigit():
                    print("Amount must be numeric.")
                    return
                e["amount_ml"] = int(new_amt)
            new_note = input(f"New note (enter to keep '{e.get('note','')}'): ").strip()
            if new_note != "":
                e["note"] = new_note
            write_entries(entries)
            print("Entry updated.")
            return
    print("Entry id not found.")

def delete_entry():
    entries = read_entries()
    eid = input("Enter entry id to delete: ").strip()
    if not eid.isdigit():
        print("Invalid id.")
        return
    eid = int(eid)
    new = [e for e in entries if e["id"] != eid]
    if len(new) == len(entries):
        print("ID not found.")
    else:
        write_entries(new)
        print(f"Deleted entry id {eid}.")

def set_goal():
    profile = load_profile()
    val = input(f"Enter daily goal in ml (current {profile.get('daily_goal_ml',2000)}): ").strip()
    if not val.isdigit():
        print("Please enter a numeric value.")
        return
    profile["daily_goal_ml"] = int(val)
    save_profile(profile)
    print("Goal updated.")

def summary_range():
    s = input("Start date (YYYY-MM-DD): ").strip()
    t = input("End date (YYYY-MM-DD): ").strip()
    try:
        sd = datetime.strptime(s, "%Y-%m-%d").date()
        ed = datetime.strptime(t, "%Y-%m-%d").date()
    except:
        print("Invalid dates.")
        return
    entries = read_entries()
    day_totals = {}
    for e in entries:
        d = e["date"]
        day_totals[d] = day_totals.get(d, 0) + e["amount_ml"]
    cur = sd
    while cur <= ed:
        print(cur.strftime("%Y-%m-%d"), "-", day_totals.get(cur.strftime("%Y-%m-%d"), 0), "ml")
        cur = cur.fromordinal(cur.toordinal() + 1)

def export_csv():
    dest = input("Export filename (e.g., export.csv): ").strip()
    if not dest:
        print("Invalid filename.")
        return
    with open(ENTRIES_FILE, "r") as src:
        data = src.read()
    with open(dest, "w") as dst:
        dst.write(data)
    print("Exported to", dest)

# ----------------- Menu -----------------

def main_menu():
    ensure_files()
    while True:
        print("\nSimpleHydrate — menu")
        print("1) Add entry")
        print("2) List today's entries")
        print("3) List entries for date")
        print("4) Edit entry")
        print("5) Delete entry")
        print("6) Set daily goal")
        print("7) Summary for date range")
        print("8) Export CSV")
        print("9) Quit")
        choice = input("Choose (1-9): ").strip()
        if choice == "1":
            add_entry()
        elif choice == "2":
            list_entries()
        elif choice == "3":
            d = input("Date (YYYY-MM-DD): ").strip()
            list_entries(d)
        elif choice == "4":
            edit_entry()
        elif choice == "5":
            delete_entry()
        elif choice == "6":
            set_goal()
        elif choice == "7":
            summary_range()
        elif choice == "8":
            export_csv()
        elif choice == "9":
            print("Goodbye — remember to hydrate! :)")
            break
        else:
            print("Invalid choice.")

if __name__ == "__main__":
    main_menu()
```
